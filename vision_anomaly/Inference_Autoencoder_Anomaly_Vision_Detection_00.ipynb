```
#19370

def creat_images_folder(videopath,image_size=(720,480),save_video=False):
    
    if save_video==True:
       fourcc = cv2.VideoWriter_fourcc('m', 'p', '4', 'v')
       #out = cv2.VideoWriter('output.mp4', fourcc, 30.0,(720,250))
       out = cv2.VideoWriter('output.mp4', fourcc, 30.0,(1920,540))
        
    vid = cv2.VideoCapture(videopath)
    i=0
    num_frames=19370
    train_size=0.8967475477542591
    while vid.isOpened():
         ret, frame = vid.read()
         if ret :
            if i< int(num_frames*train_size) :
               #frame=frame[0:250,:,:]
               #print(frame.shape)
               frame=frame[0:540,:,:]
               resized_img=cv2.resize(frame, image_size)
               cv2.imwrite("train/frame-" + str(format(i, '05d')) + ".jpg",resized_img)
               if save_video==True:
                  out.write(frame)

            elif i>=int(num_frames*train_size) and i<int(num_frames*(train_size+(1.0-train_size)/2)) :
               frame=frame[0:250,:,:]
               resized_img=cv2.resize(frame, image_size)
               cv2.imwrite("val/frame-" + str(format(i, '05d')) + ".jpg",resized_img)
            else:
               frame=frame[0:250,:,:]
               resized_img=cv2.resize(frame, image_size)
               cv2.imwrite("test/frame-" + str(format(i, '05d')) + ".jpg",resized_img)                 

            i+=1
            print("frame: ",i)
         else: 
           break
    if save_video==True:
       out.release()       
    vid.release()
```

```
import os
import torch
from torchvision import datasets, transforms
from torch.utils.data import Dataset, DataLoader

img_transforms = transforms.Compose([ 
         transforms.ToTensor()
         ])

class Images_Dataset(Dataset):
    """Images_Dataset."""

    def __init__(self,image_folder,image_list,transform=None,sequence_len=5):
        self.image_list=image_list
        self.sequence_len = sequence_len
        self.image_folder=image_folder
        self.transform=transform
    def __len__(self):
        return int(len(self.image_list)-self.sequence_len)

    def __getitem__(self, idx):
        image_sequence=[]
        for image_ in self.image_list[idx:idx+self.sequence_len]:
            img = cv2.imread(self.image_folder+image_)
            if self.transform != None:
               sample = self.transform(image=img)
               img=sample['image']
            image_tensor=img_transforms(img)
            image_sequence.append(image_tensor)

        concat_images=torch.cat(image_sequence, dim=0)
        
        

        return concat_images
```

```
import torch.nn as nn
import torch.nn.functional as F

# define the NN architecture
class ConvAutoencoder(nn.Module):
    def __init__(self,in_num_frames,out_num_frames):
        super(ConvAutoencoder, self).__init__()
        ## encoder layers ##
        # conv layer (depth from 15 --> 32), 3x3 kernels
        self.conv1 = nn.Conv2d(in_num_frames,32, 3, padding=1)
        self.bn1=nn.BatchNorm2d(32)  

        # conv layer (depth from 32 --> 64), 3x3 kernels
        self.conv2 = nn.Conv2d(32,64, 3, padding=1)
        self.bn2=nn.BatchNorm2d(64)  

        # conv layer (depth from 64 --> 128), 3x3 kernels
        self.conv3 = nn.Conv2d(64,128, 3, padding=1)        
        self.bn3=nn.BatchNorm2d(128)  
        
        # pooling layer to reduce x-y dims by two; kernel and stride of 2
        self.pool = nn.MaxPool2d(3,padding=1,stride=2)
        
        ## decoder layers ##
        ## a kernel of 2 and a stride of 2 will increase the spatial dims by 2
        self.t_conv1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.t_bn1=nn.BatchNorm2d(64)  
        
        self.t_conv2 = nn.ConvTranspose2d(64,32, 2, stride=2)
        self.t_bn2=nn.BatchNorm2d(32)  

        self.t_conv3 = nn.ConvTranspose2d(32,out_num_frames, 2, stride=2)


    def forward(self, x):
        ## encode ##
        # add hidden layers with relu activation function
        # and maxpooling after
        x = F.relu(self.conv1(x))
        x=self.bn1(x)
        #print(x.shape)
        x = self.pool(x)
        #print(x.shape)

        # add second hidden layer
        x = F.relu(self.conv2(x))
        x=self.bn2(x)
        #print(x.shape)

        x = self.pool(x)  
        #print(x.shape)

        x = F.relu(self.conv3(x))
        x=self.bn3(x)

        #print(x.shape)

        x = self.pool(x)  # compressed representation
        #print(x.shape)

        ## decode ##
        # add transpose conv layers, with relu activation function
        x = F.relu(self.t_conv1(x))
        #print(x.shape)
        x=self.t_bn1(x)
        # output layer (with sigmoid for scaling from 0 to 1)
        x = F.relu(self.t_conv2(x))
        x=self.t_bn2(x)

        #print(x.shape)

        x = self.t_conv3(x)
        #print(x.shape)
                
        return x
```
