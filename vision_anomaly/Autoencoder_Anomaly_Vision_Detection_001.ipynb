```
def creat_images_folder(videopath,image_size=(720,480)):
    vid = cv2.VideoCapture(videopath)
    i=0
    while vid.isOpened():
         ret, frame = vid.read()
         if ret :
           if videopath=="/content/video_02.mp4":
              frame=frame[30:1045,0:1890,:]
              resized_img=cv2.resize(frame, image_size)
              cv2.imwrite("test/frame-" + str(format(i, '05d')) + ".jpg",resized_img)
           elif videopath=="/content/video_00.mp4" or videopath=="/content/video_000.mp4":
              frame=frame[40:1040,:,:]
              resized_img=cv2.resize(frame, image_size)
              cv2.imwrite("train/frame-" + str(format(i, '05d')) + ".jpg",resized_img)
           i+=1
           print("frame: ",i)
         else: 
           break
    vid.release()
         
```

```
import os
import torch
from torchvision import datasets, transforms
from torch.utils.data import Dataset, DataLoader

img_transforms = transforms.Compose([ 
         transforms.ToTensor()
         ])

class Images_Dataset(Dataset):
    """Images_Dataset."""

    def __init__(self,image_folder,image_list,transform=None,sequence_len=5):
        self.image_list=image_list
        self.sequence_len = sequence_len
        self.image_folder=image_folder
        self.transform=transform
    def __len__(self):
        return int(len(self.image_list)-self.sequence_len)

    def __getitem__(self, idx):
        image_sequence=[]
        for image_ in self.image_list[idx:idx+self.sequence_len]:
            img = cv2.imread(self.image_folder+image_)
            if self.transform != None:
               sample = self.transform(image=img)
               img=sample['image']
            image_tensor=img_transforms(img)
            image_sequence.append(image_tensor)

        concat_images=torch.cat(image_sequence, dim=0)
        
        

        return concat_images
```

```

import torch.nn as nn
import torch.nn.functional as F

# define the NN architecture
class ConvAutoencoder(nn.Module):
    def __init__(self):
        super(ConvAutoencoder, self).__init__()
        ## encoder layers ##
        # conv layer (depth from 15 --> 32), 3x3 kernels
        self.conv1 = nn.Conv2d(30,32, 3, padding=1)
        self.bn1=nn.BatchNorm2d(32)  

        # conv layer (depth from 32 --> 64), 3x3 kernels
        self.conv2 = nn.Conv2d(32,64, 3, padding=1)
        self.bn2=nn.BatchNorm2d(64)  

        # conv layer (depth from 64 --> 128), 3x3 kernels
        self.conv3 = nn.Conv2d(64,128, 3, padding=1)        
        self.bn3=nn.BatchNorm2d(128)  
        
        # pooling layer to reduce x-y dims by two; kernel and stride of 2
        self.pool = nn.MaxPool2d(3,padding=1,stride=2)
        
        ## decoder layers ##
        ## a kernel of 2 and a stride of 2 will increase the spatial dims by 2
        self.t_conv1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.t_bn1=nn.BatchNorm2d(64)  
        
        self.t_conv2 = nn.ConvTranspose2d(64,32, 2, stride=2)
        self.t_bn2=nn.BatchNorm2d(32)  

        self.t_conv3 = nn.ConvTranspose2d(32,30, 2, stride=2)


    def forward(self, x):
        ## encode ##
        # add hidden layers with relu activation function
        # and maxpooling after
        x = F.relu(self.conv1(x))
        x=self.bn1(x)
        #print(x.shape)
        x = self.pool(x)
        #print(x.shape)

        # add second hidden layer
        x = F.relu(self.conv2(x))
        x=self.bn2(x)
        #print(x.shape)

        x = self.pool(x)  
        #print(x.shape)

        x = F.relu(self.conv3(x))
        x=self.bn3(x)

        #print(x.shape)

        x = self.pool(x)  # compressed representation
        #print(x.shape)

        ## decode ##
        # add transpose conv layers, with relu activation function
        x = F.relu(self.t_conv1(x))
        #print(x.shape)
        x=self.t_bn1(x)
        # output layer (with sigmoid for scaling from 0 to 1)
        x = F.relu(self.t_conv2(x))
        x=self.t_bn2(x)

        #print(x.shape)

        x = self.t_conv3(x)
        #print(x.shape)
                
        return x
```

```
import torch.nn as nn
import torch.nn.functional as F

# define the NN architecture
class ConvAutoencoder(nn.Module):
    def __init__(self,in_num_frames,out_num_frames):
        super(ConvAutoencoder, self).__init__()
        ## encoder layers ##
        # conv layer (depth from 15 --> 32), 3x3 kernels
        self.conv1 = nn.Conv2d(in_num_frames,32, 3, padding=1)
        self.bn1=nn.BatchNorm2d(32)  

        # conv layer (depth from 32 --> 64), 3x3 kernels
        self.conv2 = nn.Conv2d(32,64, 3, padding=1)
        self.bn2=nn.BatchNorm2d(64)  

        # conv layer (depth from 64 --> 128), 3x3 kernels
        self.conv3 = nn.Conv2d(64,128, 3, padding=1)        
        self.bn3=nn.BatchNorm2d(128)  
        
        # pooling layer to reduce x-y dims by two; kernel and stride of 2
        self.pool = nn.MaxPool2d(3,padding=1,stride=2)
        
        ## decoder layers ##
        ## a kernel of 2 and a stride of 2 will increase the spatial dims by 2
        self.t_conv1 = nn.ConvTranspose2d(128, 64, 2, stride=2)
        self.t_bn1=nn.BatchNorm2d(64)  
        
        self.t_conv2 = nn.ConvTranspose2d(64,32, 2, stride=2)
        self.t_bn2=nn.BatchNorm2d(32)  

        self.t_conv3 = nn.ConvTranspose2d(32,out_num_frames, 2, stride=2)


    def forward(self, x):
        ## encode ##
        # add hidden layers with relu activation function
        # and maxpooling after
        x = F.relu(self.conv1(x))
        x=self.bn1(x)
        #print(x.shape)
        x = self.pool(x)
        #print(x.shape)

        # add second hidden layer
        x = F.relu(self.conv2(x))
        x=self.bn2(x)
        #print(x.shape)

        x = self.pool(x)  
        #print(x.shape)

        x = F.relu(self.conv3(x))
        x=self.bn3(x)

        #print(x.shape)

        x = self.pool(x)  # compressed representation
        #print(x.shape)

        ## decode ##
        # add transpose conv layers, with relu activation function
        x = F.relu(self.t_conv1(x))
        #print(x.shape)
        x=self.t_bn1(x)
        # output layer (with sigmoid for scaling from 0 to 1)
        x = F.relu(self.t_conv2(x))
        x=self.t_bn2(x)

        #print(x.shape)

        x = self.t_conv3(x)
        #print(x.shape)
                
        return x
```

```
from tqdm import tqdm

def training(model, train_loader,in_num_frames, Epochs):
    model.train()
    train_loss = []
    #for epoch in range(Epochs):
    running_loss = 0.0
    for input_ in tqdm(train_loader,position=0, leave=True):
        input_ = input_.to(device)
        optimizer.zero_grad()
        outputs = model(input_[:,0:in_num_frames:,:]).to(device)
        
        loss = criterion(outputs, input_)
        
        #loss=ssim_loss(input_.detach(),outputs)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()

    loss = running_loss / len(train_loader)
    #train_loss.append(loss)
    #print('Epoch {} of {}, Train Loss: {:.3f}'.format(
    #    epoch+1, Epochs, loss))

        

    return loss
```

```
def evaluate(model, data_loader,in_num_frames):
    model.eval()
    loss_list=[]
    with torch.no_grad():
 
      for input_ in tqdm(data_loader,position=0, leave=True):
        input_ = input_.to(device)
        optimizer.zero_grad()
        outputs = model(input_[:,0:in_num_frames:,:].detach()).to(device)
        loss = criterion(outputs, input_)
        loss_list.append(loss.item())
    return  loss_list
```

```
import numpy as np
def train_model(model,train_loader,val_loader,in_num_frames,Epochs,model_name):
    best_validation_loss=np.inf

    for epoch in range(Epochs):
        train_loss=training(model, train_loader,in_num_frames,Epochs)
        val_loss=evaluate(model,val_loader,in_num_frames)
    
        print('Epoch {} of {}, Train Loss: {:.5f}, Val Loss: {:.5f}'.format(epoch+1, Epochs, train_loss,np.mean(val_loss)))
        if np.mean(val_loss) < best_validation_loss:
            best_validation_loss=np.mean(val_loss)
            torch.save(model.state_dict(),'/content/drive/MyDrive/Anomaly vision detection/'+model_name+'.pt')
```
